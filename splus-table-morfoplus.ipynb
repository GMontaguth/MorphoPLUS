```python
import pandas as pd
from pylegs.io import *


df=read_table('/media/gissel/Expansion/SCALE/Paquetes/A1644.parquet')


# === 2. Definir las columnas que realmente existen en tu df ===
cols_keep = [
    'z', 'ra', 'dec',
    'FLUX_RADIUS_50', 'FLUX_RADIUS_90',
    'r_auto', 'e_r_auto',
    'u_auto', 'e_u_auto',
    'J0378_auto', 'e_J0378_auto',
    'J0395_auto', 'e_J0395_auto',
    'J0410_auto', 'e_J0410_auto',
    'J0430_auto', 'e_J0430_auto',
    'g_auto', 'e_g_auto',
    'J0515_auto', 'e_J0515_auto',
    'J0660_auto', 'e_J0660_auto',
    'i_auto', 'e_i_auto',
    'z_auto', 'e_z_auto',
    'J0861_auto', 'e_J0861_auto',
    'Field'
]

# Nos quedamos solo con esas columnas
df_reduced = df[cols_keep].copy()

# === 3. Crear la columna ID tipo A119_1, A119_2, ... ===
df_reduced['ID'] = [f"A119_{i+1}" for i in range(len(df_reduced))]

# === 4. Crear columnas RA y DEC a partir de ra, dec ===
df_reduced['RA'] = df_reduced['ra']
df_reduced['DEC'] = df_reduced['dec']


# === 6. (Opcional) Calcular ELONGATION si tenés A y B en el df original ===
if ('A' in df.columns) and ('B' in df.columns):
    df_reduced['ELONGATION'] = df['A'] / df['B']
else:
    df_reduced['ELONGATION'] = pd.NA

# Copiar THETA y ELLIPTICITY si existen
if 'THETA' in df.columns:
    df_reduced['THETA'] = df['THETA']
else:
    df_reduced['THETA'] = pd.NA

if 'ELLIPTICITY' in df.columns:
    df_reduced['ELLIPTICITY'] = df['ELLIPTICITY']
else:
    df_reduced['ELLIPTICITY'] = pd.NA

# === 7. Reordenar columnas para que queden parecido a tu formato estándar ===
cols_final = [
    'z', 'ra', 'dec',
    'ID', 'RA', 'DEC', 
    'ELONGATION', 'THETA', 'ELLIPTICITY',
    'FLUX_RADIUS_50', 'FLUX_RADIUS_90',
    'r_auto', 'e_r_auto',
    'u_auto', 'e_u_auto',
    'J0378_auto', 'e_J0378_auto',
    'J0395_auto', 'e_J0395_auto',
    'J0410_auto', 'e_J0410_auto',
    'J0430_auto', 'e_J0430_auto',
    'g_auto', 'e_g_auto',
    'J0515_auto', 'e_J0515_auto',
    'J0660_auto', 'e_J0660_auto',
    'i_auto', 'e_i_auto',
    'z_auto', 'e_z_auto',
    'J0861_auto', 'e_J0861_auto',
    'Field'
]

df_reduced = df_reduced[cols_final]

# === 8. Guardar el resultado ===
df_reduced.to_csv('Catalogos/SPLUS_Table.csv', index=False)

print("✅ Archivo 'SPLUS_Table_A119.csv' guardado con éxito.")
```

```python
import pandas as pd

SRC = "1158_lit_0ct_21.csv"
OUT = "SPLUS_Table.csv"

df = pd.read_csv(SRC)

def pick(df_cols, *candidates):
    for c in candidates:
        if c in df_cols:
            return c
    return None

# ==== Definimos alias por cada columna destino ====
alias = {
    # básicos
    'z'                  : ['z'],
    'ra'                 : ['ra','RA'],  # crudo por si lo querés conservar
    'dec'                : ['dec','DEC'],
    'ID'                 : ['id','ID'],
    'Field'              : ['field','Field'],

    # coordenadas finales que querés en salida
    'RA'                 : ['input_RA','RA'],
    'DEC'                : ['input_DEC','DEC'],

    # métricas geométricas/píxel
    'X'                  : ['x_pixel_det','X_IMAGE'],
    'Y'                  : ['y_pixel_det','Y_IMAGE'],
    'THETA'              : ['theta_det','THETA','THETA_IMAGE'],
    'ELONGATION'         : ['elongation_det','ELONGATION'],
    'ELLIPTICITY'        : ['ellipticity_det','ELLIPTICITY'],

    # radios de flujo (preferido: tus columnas; fallback: banda r)
    'FLUX_RADIUS_50'     : ['flux_radius_as_50_det','FLUX_RADIUS_50','FLUX_RADIUS_50_R'],
    'FLUX_RADIUS_90'     : ['flux_radius_as_90_det','FLUX_RADIUS_90','FLUX_RADIUS_90_R'],

    # fotometría AUTO + errores (viejo: mag_auto_* / err_mag_auto_*; nuevo: *_AUTO / e_*_AUTO)
    'r_auto'             : ['mag_auto_r','R_AUTO','r_auto'],
    'e_r_auto'           : ['err_mag_auto_r','e_R_AUTO','e_r_auto'],

    'u_auto'             : ['mag_auto_u','U_AUTO','u_auto'],
    'e_u_auto'           : ['err_mag_auto_u','e_U_AUTO','e_u_auto'],

    # filtros estrechos S-PLUS → prefijos Fxxx en “versión nueva”
    'J0378_auto'         : ['mag_auto_j0378','F378_AUTO','J0378_AUTO','J0378_auto'],
    'e_J0378_auto'       : ['err_mag_auto_j0378','e_F378_AUTO','e_J0378_AUTO','e_J0378_auto'],

    'J0395_auto'         : ['mag_auto_j0395','F395_AUTO','J0395_AUTO','J0395_auto'],
    'e_J0395_auto'       : ['err_mag_auto_j0395','e_F395_AUTO','e_J0395_AUTO','e_J0395_auto'],

    'J0410_auto'         : ['mag_auto_j0410','F410_AUTO','J0410_AUTO','J0410_auto'],
    'e_J0410_auto'       : ['err_mag_auto_j0410','e_F410_AUTO','e_J0410_AUTO','e_J0410_auto'],

    'J0430_auto'         : ['mag_auto_j0430','F430_AUTO','J0430_AUTO','J0430_auto'],
    'e_J0430_auto'       : ['err_mag_auto_j0430','e_F430_AUTO','e_J0430_AUTO','e_J0430_auto'],

    'g_auto'             : ['mag_auto_g','G_AUTO','g_auto'],
    'e_g_auto'           : ['err_mag_auto_g','e_G_AUTO','e_g_auto'],

    'J0515_auto'         : ['mag_auto_j0515','F515_AUTO','J0515_AUTO','J0515_auto'],
    'e_J0515_auto'       : ['err_mag_auto_j0515','e_F515_AUTO','e_J0515_AUTO','e_J0515_auto'],

    'J0660_auto'         : ['mag_auto_j0660','F660_AUTO','J0660_AUTO','J0660_auto'],
    'e_J0660_auto'       : ['err_mag_auto_j0660','e_F660_AUTO','e_J0660_AUTO','e_J0660_auto'],

    'i_auto'             : ['mag_auto_i','I_AUTO','i_auto'],
    'e_i_auto'           : ['err_mag_auto_i','e_I_AUTO','e_i_auto'],

    'z_auto'             : ['mag_auto_z','Z_AUTO','z_auto'],
    'e_z_auto'           : ['err_mag_auto_z','e_Z_AUTO','e_z_auto'],

    'J0861_auto'         : ['mag_auto_j0861','F861_AUTO','J0861_AUTO','J0861_auto'],
    'e_J0861_auto'       : ['err_mag_auto_j0861','e_F861_AUTO','e_J0861_AUTO','e_J0861_auto'],
}

# ==== Construimos el diccionario {col_origen: col_destino} solo con las que existan ====
src_to_dst = {}
missing = []

for dst, candidates in alias.items():
    src = pick(df.columns, *candidates)
    if src is not None:
        src_to_dst[src] = dst
    else:
        missing.append((dst, candidates))

# Extra: mantenemos también explícitamente 'z','ra','dec' si existen (tus “raw”)
# (ya queda cubierto arriba, pero lo dejo por claridad)

# ==== Selección + renombre ====
df_reduced = df[list(src_to_dst.keys())].rename(columns=src_to_dst)

# ==== Orden sugerido de salida (si falta alguno, se ignora) ====
order = [
    'z','ra','dec','ID','Field',
    'RA','DEC','X','Y','THETA','ELONGATION','ELLIPTICITY',
    'FLUX_RADIUS_50','FLUX_RADIUS_90',
    'r_auto','e_r_auto','u_auto','e_u_auto',
    'J0378_auto','e_J0378_auto','J0395_auto','e_J0395_auto',
    'J0410_auto','e_J0410_auto','J0430_auto','e_J0430_auto',
    'g_auto','e_g_auto','J0515_auto','e_J0515_auto',
    'J0660_auto','e_J0660_auto','i_auto','e_i_auto',
    'z_auto','e_z_auto','J0861_auto','e_J0861_auto'
]
df_reduced = df_reduced[[c for c in order if c in df_reduced.columns]]

# ==== Guardar ====
df_reduced.to_csv(OUT, index=False)
print(f"✅ Archivo '{OUT}' guardado con éxito.")
if missing:
    print("\nAviso: no se encontraron estas columnas destino (con sus alias probados):")
    for dst, cand in missing:
        print(f"  - {dst}: {cand}")
```
